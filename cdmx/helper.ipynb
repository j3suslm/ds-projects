```python
import polars as pl
```

```python
data = (
    pl.scan_parquet('data/fgj_files.parquet')
        .with_columns(
            pl.col("fecha_hecho").str.extract(r"(\d{4}-\d{2}-\d{2})").alias("fecha_hecho")
        )
        .with_columns(
            pl.col('fecha_hecho').str.to_date().alias('fecha_hecho')
        )
        .collect()
)
```

```python
df = (
    data
        .select('fecha_hecho','delito','latitud','longitud')
        .filter(pl.col('delito').str.contains('ROBO A TRANSEUNTE'))
        .drop_nulls()
)
```

```python
from datetime import datetime

df.filter(pl.col('fecha_hecho') >= datetime(2014,1,1))
```
